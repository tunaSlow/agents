# Three Parts to this lab

## Part 1: A simple "Agent" and "Agent Loop"

Basically an LLM call. We'll add tracing and streaming to the mix.

## Part 2: Adding a Tool

A familiar one, but oh-so-easy

## Part 3: Adding Memory

So that different Agent calls know about each other

```python
import asyncio
import os
from dotenv import load_dotenv
from openai import AsyncOpenAI
from agents import Agent, Runner, function_tool, set_tracing_disabled
from agents.models.openai_chatcompletions import OpenAIChatCompletionsModel
import nest_asyncio


# 1. Environment Setup
load_dotenv()
set_tracing_disabled('disabled')  # Prevents telemetry calls to OpenAI endpoints

# 2. Configure Groq-backed AsyncOpenAI Client
groq_client = AsyncOpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

# 3. Instantiate OpenAI ChatCompletions Adapter for Groq
groq_model = OpenAIChatCompletionsModel(
    model="openai/gpt-oss-120b",
    openai_client=groq_client
)

# 4. Define Agent Tool
@function_tool
def analyze_log_file(filename: str) -> str:
    """Analyzes a log file for runtime exceptions and error spikes."""
    return f"Log file '{filename}' analyzed: 0 fatal errors, 2 minor warnings detected."

# 5. Define Groq Agent
ops_agent = Agent(
    name="DevOps Assistant",
    instructions="You diagnose system logs using available tool execution loops.",
    model=groq_model,
    tools=[analyze_log_file]
)

# 6. Execute Workflow
async def main():
    result = await Runner.run(ops_agent, "Check system logs in 'app.log' for issues.")
    print("Agent Response:\n", result.final_output)

if __name__ == "__main__":
    nest_asyncio.apply()
    asyncio.run(main())
```
