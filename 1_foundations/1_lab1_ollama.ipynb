```python
# !ollama pull llama3.2

from openai import OpenAI

OLLAMA_BASE_URL = "http://localhost:11434/v1"
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key="anything")
question1 = "what is 2 + 45?"
response = ollama.chat.completions.create(model="llama3.2", messages=[{"role":"user", "content": question1}])
answer = response.choices[0].message.content
print(answer)
```

```python
message = f"""
Here is a question:
{question1}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages=[{"role":"user", "content": message}]
response = ollama.chat.completions.create(model="llama3.2", messages=messages)
answer = response.choices[0].message.content
print(answer)
```
